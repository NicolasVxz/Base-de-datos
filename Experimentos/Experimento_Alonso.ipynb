```python
import time

def medir_carga(ruta, repeticiones=3):
    tiempos = []
    for i in range(repeticiones):
        inicio = time.perf_counter()
        df = spark.read.csv(ruta, header=True, inferSchema=True, sep=",")
        filas = df.count()
        fin = time.perf_counter()
        tiempos.append(round(fin - inicio, 3))
        print(f"Repetición {i+1}: {filas} filas en {tiempos[-1]} s")
    print("Promedio:", round(sum(tiempos) / len(tiempos), 3), "s")
    return tiempos
```

```python
t200 = medir_carga("/Volumes/workspace/experimento_alonso/ex/EFF/archivosAtencionCita.csv")
```

```python
t930 = medir_carga("/Volumes/workspace/experimento_alonso/ex/EFF/AtencionCita.csv")
```

```python
import pandas as pd

prom400 = round(sum(t200) / len(t200), 3)
prom999 = round(sum(t930) / len(t930), 3)

tabla = pd.DataFrame({
    "Archivo": ["200 filas", "930 filas"],
    "Repeticion_1": [t200[0], t930[0]],
    "Repeticion_2": [t200[1], t930[1]],
    "Repeticion_3": [t200[2], t930[2]],
    "Promedio_s": [prom400, prom999],
})

dfTabla = spark.createDataFrame(tabla)
display(dfTabla)
```
