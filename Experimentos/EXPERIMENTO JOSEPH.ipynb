```python
dftablapago = spark.read.csv("dbfs:/Workspace/CLINICA/BD_CLINICA/TABLA_PAGOS_CITAS_AUMENTADO.csv",header=True,inferSchema=True,sep=",")
```

```python
display(dftablapago.describe())
```

```python
dftablapago = (dftablapago
  .withColumnRenamed("Año1", "AÑO")
  .withColumnRenamed("Trimestre2", "TRIMESTRE")
  .withColumnRenamed("Mes3", "MES")
  .withColumnRenamed("Día4", "DIS")
  .withColumnRenamed("Suma de DESCT PORCENTAJE", "DESCT_PORCENTAJE")
  .withColumnRenamed("FINALIZACION ESTADO", "FIN_ESTADO")
  .withColumnRenamed("Suma de MONTO", "MONTO_INICIAL")
  .withColumnRenamed("Suma de PAGO_FINAL", "PAGO_FINAL")
  )

display(dftablapago)
```

```python
dftablapago =  dftablapago.drop("Año5","Trimestre6","Mes7","Día8")
```

```python
display(dftablapago)
```

```python
dftablapago.write.savedAsTable("tb_pago_cita")
```
