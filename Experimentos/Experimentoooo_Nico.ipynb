```python
import time

def medir_carga(ruta, repeticiones=3):
    tiempos = []
    for i in range(repeticiones):
        inicio = time.perf_counter()
        df = spark.read.csv(ruta, header=True, inferSchema=True, sep=",")
        filas = df.count()
        fin = time.perf_counter()
        tiempos.append(round(fin - inicio, 3))
        print(f"Repetición {i+1}: {filas} filas en {tiempos[-1]} s")
    print("Promedio:", round(sum(tiempos) / len(tiempos), 3), "s")
    return tiempos
```

```python
t400 = medir_carga("/Volumes/workspace/experimentooo/experimentoooo/EXPERIMENTO/Reclamo_400_filas.csv")
```

```python
t999 = medir_carga("/Volumes/workspace/experimentooo/experimentoooo/EXPERIMENTO/Tabla_1000.csv")
```

```python
import pandas as pd

prom400 = round(sum(t400) / len(t400), 3)
prom999 = round(sum(t999) / len(t999), 3)

tabla = pd.DataFrame({
    "Archivo": ["400 filas", "999 filas"],
    "Repeticion_1": [t400[0], t999[0]],
    "Repeticion_2": [t400[1], t999[1]],
    "Repeticion_3": [t400[2], t999[2]],
    "Promedio_s": [prom400, prom999],
})

dfTabla = spark.createDataFrame(tabla)
display(dfTabla)
```
