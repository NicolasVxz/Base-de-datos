```python
import time
def medir_carga(ruta, repeticiones=3):
    tiempos = []
    for i in range(repeticiones):
        inicio = time.perf_counter()
        df = spark.read.csv(ruta, header=True, inferSchema=True, sep=",")
        filas = df.count()
        fin = time.perf_counter()
        tiempos.append(round(fin - inicio, 3))
        print(f"Repetición {i+1}: {filas} filas en {tiempos[-1]} s")
    print("Promedio:", round(sum(tiempos) / len(tiempos), 3), "s")
    return tiempos
```

```python
t100=medir_carga("/Volumes/workspace/experimento/experimentoyerson/yerson/pagofarmaxp.csv")
```

```python
t861=medir_carga("/Volumes/workspace/experimento/experimentoyerson/yerson/PagoFarma2.csv")
```

```python
import pandas as pd
prom100 = round(sum(t100) / len(t100), 3)
prom861 = round(sum(t861) / len(t861), 3)
tabla = pd.DataFrame({
    "Archivo": ["100 filas", "861 filas"],
    "Repeticion_1": [t100[0], t861[0]],
    "Repeticion_2": [t100[1], t861[1]],
    "Repeticion_3": [t100[2], t861[2]],
    "Promedio_s": [prom100, prom861],
})
dfTabla = spark.createDataFrame(tabla)
display(dfTabla)
```
